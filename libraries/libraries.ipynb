```python
number = 27

'''Square root
Floor of 27.8
Ceiling of 27.2
5 factorial
GCD of 48 and 18
LCM of 4 and 6
Value of π

Use the math library.'''

import math

print(math.sqrt(number))
print(math.floor(27.8))
print(math.ceil(27.2))
print(math.factorial(5))
print(math.gcd(48, 18))
print(math.lcm(4, 6))
print(math.pi)
print(math.e)
```

```python
# Question 2 — Circle calculator
#
# Given:
#
# radius = 7
#
# Using math.pi, calculate and print:
#
# Area of the circle
# Circumference of the circle
#
# Formula:
#
# Area = π × r²
# Circumference = 2 × π × r
#
# Write the Python code yourself.

import math

r = 7
area = math.pi * math.pow(r, 2)
circumference = 2 * math.pi * r

print("Area of Circle is : ", area)
print("Circumference of Circle is : ", circumference)
```

```python
# Your task:
#
# Get today's date.
# Create a timedelta of 10 days.
# Add it to today's date.
# Print the resulting date.
#
# Don't look up the syntax. Think about what timedelta represents: a duration.


from datetime import date, timedelta

date = date.today()

duration = timedelta(days = 10)

result = date + duration
print(result)
```

```python
# Now test subtraction
#
# This time, calculate the date 15 days before today.
#
# Use timedelta and subtraction.
#
# Don't use any new function. Just modify the logic you already used.

from datetime import date, timedelta

today = date.today()

duration = timedelta(days = 15)

result = today - duration
print(result)
```

```python
# Question 4 — Difference between two dates
#
# Now let's test the other important use of timedelta.
#
# Suppose:
#
# start = date(2026, 9, 1)
# end = date(2026, 9, 30)
#
# Calculate how many days are between start and end.
#
# Hint: You don't need to create a timedelta yourself this time. Subtract one date from the other.

from datetime import date, time

start = date(2026, 9, 1)
end = date(2026, 9, 30)

print(end - start)

```

```python
# Question 5 — datetime + timedelta
#
# Let's make this slightly more practical.
#
# Suppose a user logs in at the current time and their session expires 2 hours later.
#
# Write code that:
#
# Imports datetime and timedelta
# Gets the current datetime
# Creates a 2-hour duration
# Calculates the expiry time
# Prints both the current time and expiry time
#
# This is a pattern you'll encounter in backend development.

from datetime import datetime, timedelta

current = datetime.now()
duration = timedelta(hours = 2)

expiry_time = current + duration
print("Current Time :", current)
print("Expiry Time :", expiry_time)
```
