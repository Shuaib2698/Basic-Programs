```python
numbers = [10, 20, 30, 20, 40, 10]
# numbers = [5, 8, 2, 5, 8]
def first_duplicate(numbers):
    num = []

    for i in numbers:
        if i in num:
            return i
        else:
            num.append(i)
    return None

print(first_duplicate(numbers))
```

```python
# numbers = [10, 50, 50, 40, 30]
numbers = [10, 10, 10]
def second_largest(numbers):
    largest = float('-inf')
    second = float('-inf')

    for i in numbers:
        if i > largest:
            second = largest
            largest = i
        elif i > second and i < largest:
            second = i

    if second == largest or second == float('-inf'):
        return None

    return second

print(second_largest(numbers))
```
