```python
# Q22. Group Employees
# Given:
# employees = [    {"name": "Rahul", "department": "IT"},    {"name": "Aman", "department": "HR"},    {"name": "Priya", "department": "IT"},    {"name": "Sneha", "department": "Finance"},    {"name": "Arjun", "department": "HR"}]
#
# Group employees by department.
# Expected structure:
# {
#     "IT": ["Rahul", "Priya"],
#     "HR": ["Aman", "Arjun"],
#     "Finance": ["Sneha"]
# }

employees = [
    {"name": "Rahul", "department": "IT"},
    {"name": "Aman", "department": "HR"},
    {"name": "Priya", "department": "IT"},
    {"name": "Sneha", "department": "Finance"},
    {"name": "Arjun", "department": "HR"}
]

def group_employee(employees):
    result = {
        "IT" : [],
        "HR" : [],
        "Finance" : []
    }

    for emp in employees:
        department = emp.get('department')

        if department in result:
            result[department].append(emp.get("name"))

    return result

print(group_employee(employees))
```

```python
# Q26. *args
# Create:
# calculate_sum(*args)
#
# It should accept any number of numbers and return their total.
# Example:
# calculate_sum(10, 20, 30, 40)
# expected
# 100

def calculate_sum(*args):
    total = 0

    for i in args:
        total+=i
    return total

print(calculate_sum(10, 20, 30, 40))
```

```python
# Q27. **kwargs
# Create:
# display_employee(**kwargs)
#
# Call it using:
# display_employee(    name="Rahul",    age=25,    department="IT",    salary=50000)
#
# Print each key and value.

def display_employee(**kwargs):
    for key, value in kwargs.items():
        print(key, ": ", value)


display_employee(
    name="Rahul",
    age=25,
    department="IT",
    salary=50000
)
```

```python
# Q28. Combined *args + **kwargs
# Create:
# employee_details(*skills, **details)
#
# Example:
# employee_details(    "Python",    "Django",    "React",    name="Rahul",    experience=2,    salary=50000)
#
# Return/display the employee's skills and details separately.

def employee_details(*skills, **details):

    profile = {
            "Name" : details["name"],
            "skills" : skills,
            "Experience" : details["experience"],
            "salary" : details["salary"]
        }

    return profile


employee_details(
    "Python",
    "Django",
    "React",
    name="Rahul",
    experience=2,
    salary=50000
)
```

```python
# Q29. Recursive Countdown
# Write a recursive function that prints:
# 5
# 4
# 3
# 2
# 1

n = 5
def recursive_num(n):
    if n == 0:
        return

    print(n)

    recursive_num(n-1)
recursive_num(n)
```

```python
# Q30. Recursive Factorial
# Create:
# factorial(n)
#
# using recursion.
# Example:
# factorial(5)
#
# Expected:
# 120

def fact(n):
    if n == 0:
        return 0
    if n ==1:
        return 1

    return n * fact(n-1)

print(fact(5))
```

```python
# Q37. sorted()
# Given:
# employees = [    {"name": "Rahul", "salary": 45000},    {"name": "Aman", "salary": 70000},    {"name": "Priya", "salary": 55000},    {"name": "Sneha", "salary": 90000}]
#
# Sort the employees by salary in descending order using sorted() and lambda.

employees = [
    {"name": "Rahul", "salary": 45000},
    {"name": "Aman", "salary": 70000},
    {"name": "Priya", "salary": 55000},
    {"name": "Sneha", "salary": 90000}
]

employees = sorted(employees, key = lambda employee : employee["salary"], reverse = True)
print(employees)
```

```python
# Q38. Map + Dictionary Transformation
# Given:
# employees = [    {"name": "Rahul", "salary": 40000},    {"name": "Aman", "salary": 50000},    {"name": "Priya", "salary": 60000}]
#
# Use map() to create a new list where each employee's salary is increased by 10%.
# Expected structure:
# [    {"name": "Rahul", "salary": 44000},    {"name": "Aman", "salary": 55000},    {"name": "Priya", "salary": 66000}]

employees = [
    {"name": "Rahul", "salary": 40000},
    {"name": "Aman", "salary": 50000},
    {"name": "Priya", "salary": 60000}
]

# 'emp' represents each dictionary as map loops through the list
updated_employees = list(map(lambda emp: {"name": emp["name"], "salary": int(emp["salary"] * 1.10)}, employees))

print(updated_employees)
```

```python
# Q40. Employee Salary Processing
# Using the same employee data:
# 1. Filter employees whose salary is greater than 50000.
# 2. Increase their salary by 10%.
# 3. Sort them by salary in descending order.
# Use the concepts you practiced such as:
# filter()
# map()
# lambda
# sorted()
employees = [
    {"name": "Rahul", "department": "IT", "salary": 45000},
    {"name": "Aman", "department": "HR", "salary": 55000},
    {"name": "Priya", "department": "IT", "salary": 75000},
    {"name": "Sneha", "department": "Finance", "salary": 65000},
    {"name": "Arjun", "department": "IT", "salary": 85000}
]

ans = sorted(
    map(
        # Included 'department' so we don't lose that data in the final output
        lambda emp: {"name": emp["name"], "department": emp["department"], "salary": int(emp["salary"] * 1.10)},
        filter(lambda emp: emp["salary"] > 50000, employees)
    ),
    key=lambda emp: emp["salary"],
    reverse=True
)

print(ans)
```

```python
# Q39. Employee Filtering
# Given:
# employees = [    {"name": "Rahul", "department": "IT", "salary": 45000},    {"name": "Aman", "department": "HR", "salary": 55000},    {"name": "Priya", "department": "IT", "salary": 75000},    {"name": "Sneha", "department": "Finance", "salary": 65000},    {"name": "Arjun", "department": "IT", "salary": 85000}]
#
#
# Find all IT employees whose salary is greater than 60000.
employees = [
    {"name": "Rahul", "department": "IT", "salary": 45000},
    {"name": "Aman", "department": "HR", "salary": 55000},
    {"name": "Priya", "department": "IT", "salary": 75000},
    {"name": "Sneha", "department": "Finance", "salary": 65000},
    {"name": "Arjun", "department": "IT", "salary": 85000}
]

# The lambda returns True only if BOTH conditions are met
filtered_employees = list(filter(lambda emp: emp["department"] == "IT" and emp["salary"] > 60000, employees))

print(filtered_employees)
```

```python
'''Q41. Department Salary Analysis
Given:
employees = [    {"name": "Rahul", "department": "IT", "salary": 45000},    {"name": "Aman", "department": "HR", "salary": 55000},    {"name": "Priya", "department": "IT", "salary": 75000},    {"name": "Sneha", "department": "Finance", "salary": 65000},    {"name": "Arjun", "department": "IT", "salary": 85000},    {"name": "Kiran", "department": "HR", "salary": 60000}]


Create a dictionary containing the total salary for each department.
Expected structure:
{    "IT": ...,    "HR": ...,    "Finance": ...}'''
employees = [
    {"name": "Rahul", "department": "IT", "salary": 45000},
    {"name": "Aman", "department": "HR", "salary": 55000},
    {"name": "Priya", "department": "IT", "salary": 75000},
    {"name": "Sneha", "department": "Finance", "salary": 65000},
    {"name": "Arjun", "department": "IT", "salary": 85000},
    {"name": "Kiran", "department": "HR", "salary": 60000}
]

department_totals = {}

for emp in employees:
    dept = emp["department"]
    # .get() fetches the current total (defaults to 0 if the department isn't in the dictionary yet)
    department_totals[dept] = department_totals.get(dept, 0) + emp["salary"]

print(department_totals)
```

```python
# Q42. Find Employee With Highest Salary Per Department
# Using the same data, produce:
# {    "IT": {"name": "...", "salary": ...},    "HR": {"name": "...", "salary": ...},    "Finance": {"name": "...", "salary": ...}}
#
#
# The program must calculate the result dynamically.

highest_paid_per_dept = {}

for emp in employees:
    dept = emp["department"]

    # If the department isn't tracked yet, OR the current employee earns more than the recorded one
    if dept not in highest_paid_per_dept or emp["salary"] > highest_paid_per_dept[dept]["salary"]:
        highest_paid_per_dept[dept] = {"name": emp["name"], "salary": emp["salary"]}

print(highest_paid_per_dept)
```

```python
# Q43. Two Sum
# Given:
# numbers = [2, 7, 11, 15]target = 9
#
#
# Find the indices of two numbers whose sum equals the target.
# Expected:
# [0, 1]
#
#
# Try to solve this using the dictionary approach you practiced.

numbers = [2, 7, 11, 15]
target = 9

seen = {} # Will store {number: index}

for i, num in enumerate(numbers):
    difference = target - num

    if difference in seen:
        print([seen[difference], i])
        break # Stop searching once we find the pair

    seen[num] = i
```
