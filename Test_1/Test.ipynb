```python
# Q7. Prime Numbers
# Print all prime numbers between:
# 1 and 100

def prime_nums():
    def is_Prime(n):
        if n < 2:
            return False

        for i in range(2, n):
            if n % i == 0:
                return False
        return True

    for i in range(1, 101):
        if is_Prime(i):
            print(i, end = " ")

prime_nums()
```

```python
# Q11. Character Frequency
# Given:
text = "programming"
#
# Create a dictionary containing the frequency of each character.
# Expected:
# {'p': 1, 'r': 2, 'o': 1, 'g': 2, 'a': 1, 'm': 2, 'i': 1, 'n': 1}

def char_freq(text):
    freq = {}

    for i in text:
        if i not in freq:
            freq[i] = 1
        else:
            freq[i] +=1

    return freq

print(char_freq(text))
```

```python
# Q12. Duplicate Characters
# Given:
text = "programming"
#
# Find the characters that occur more than once.
# Expected:
# r, g, m
# Order does not matter.

def duplicate_char(text):
    freq = {}

    for i in text:
        freq[i] = freq.get(i, 0) + 1

    duplicates = []

    for key, value in freq.items():
        if freq[key] >= 2:
            duplicates.append(key)
    return ", ".join(duplicates)

print(duplicate_char(text))
```

```python
# Q13. Anagram
# Check whether these two strings are anagrams:
text1 = "listen"
text2 = "silent"
#
# Do not hard-code the answer.

def anagram(text1, text2):
    if sorted(text1) == sorted(text2):
        return True
    return False

print(anagram(text1, text2))
```

```python
# Q14. List Transformation
# Given:
numbers = [2, 5, 8, 11, 14, 17, 20]
#
# Create a new list containing the squares of only the even numbers.
# Expected:
# [4, 64, 196, 400]

def square_list(numbers):
    new_list = []

    for i in numbers:
        if i % 2 == 0:
            new_list.append(i**2)

    return new_list

print(square_list(numbers))
```

```python
# Q15. Remove Duplicates
# Given:
numbers = [10, 20, 10, 30, 20, 40, 30, 50]
#
# Create a new list containing unique values.

def unique_list(numbers):
    unique = []
    seen = []

    for i in numbers:
        if i in seen:
            if i not in unique:
                unique.append(i)
        else:
            seen.append(i)

    return unique

print(unique_list(numbers))
```

```python
# Q16. Second Largest
# Given:
numbers = [45, 12, 78, 34, 89, 23, 89, 56]
#
# Find the second largest unique number.
# Expected:
# 78

def second_largest(n):
    largest = float('-inf')
    second = float('-inf')

    for i in n:
        if i > largest:
            second = largest
            largest = i

        elif i > second and i < largest:
            second = i

    return second

print(second_largest(numbers))
```

```python
# Q17. Common Elements
# Given:
# list1 = [10, 20, 30, 40, 50]list2 = [30, 40, 60, 70, 50]
#
# Find the common elements.
# Expected:
# [30, 40, 50]

l1 = [10, 20, 30, 40, 50]
l2 = [30, 40, 60, 70, 50]

def common_element(l1, l2):
    comm = []

    for i in l1:
        if i in l2:
            comm.append(i)

    return comm

print(common_element(l1, l2))
```

```python
# Q18. Flatten a List
# Given:
numbers = [[1, 2], [3, 4], [5, 6], [7, 8]]
#
# Create:
# [1, 2, 3, 4, 5, 6, 7, 8]

def flatten(n):
    flat = []

    for i in n:
        if isinstance(i, list):
            flat.extend(flatten(i))
        else:
            flat.append(i)

    return flat

print(flatten(numbers))
```

```python
# Q19. Employee Salary
# Given:
# employees = {    "Rahul": 45000,    "Aman": 55000,    "Priya": 62000,    "Sneha": 48000,    "Arjun": 70000}
#
# Print the names of employees whose salary is greater than 50000.
employees = {
    "Rahul": 45000,
    "Aman": 55000,
    "Priya": 62000,
    "Sneha": 48000,
    "Arjun": 70000
}

def emp_salary(employees):

    emp = []
    for key, value in employees.items():
        if employees[key] > 50000:
            emp.append(key)
    return emp

print(emp_salary(employees))
```
